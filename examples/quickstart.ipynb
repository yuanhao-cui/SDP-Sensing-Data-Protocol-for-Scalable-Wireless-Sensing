# WSDP Quickstart

This notebook walks through the basic workflow of the **WSDP** (Wi-Fi Sensing Data Protocol) library:

1. Install the package
2. Download and extract the elderAL sample dataset (requires a free [SDP8.org](https://sdp8.org) account)
3. Run the classification pipeline
4. Explore the model registry
5. Use algorithm / preprocessing registries

Prefer the written guide? The [documentation Quick Start](https://github.com/yuanhao-cui/SDP-Sensing-Data-Protocol-for-Scalable-Wireless-Sensing/blob/main/docs/getting-started/quickstart.md) covers the same workflow from the CLI.

```python
# Install wsdp (uncomment if running in Colab / fresh environment)
# !pip install wsdp
```

```python
import wsdp

print(f"wsdp version: {wsdp.__version__}")

# elderAL requires a free SDP8.org account — paste your JWT token below
# (or pass email="you@example.com", password="...") to run the download.
wsdp.download(
    "elderAL",
    "./data",
    token="YOUR_JWT_TOKEN",
)
```

```python
# Extract the downloaded archive (data_new.zip) so the pipeline can read the CSV files
import zipfile

with zipfile.ZipFile("./data/data_new.zip") as z:
    z.extractall("./data/elderAL")
```

```python
# Run the full classification pipeline
wsdp.pipeline(
    input_path="./data/elderAL",
    output_folder="./output",
    dataset="elderAL",
    num_seeds=1,        # use 1 seed for a quick demo
    num_epochs=5,       # few epochs for demonstration
)
```

## Available Models

WSDP ships with a rich model registry spanning baselines, mainstream architectures,
state-of-the-art models, and specialised WiFi CSI models.

```python
from wsdp.models import list_models

# List every registered model
for name in list_models():
    print(name)
```

## Custom Preprocessing

You can inspect and extend the algorithm / processor registries to plug in
your own preprocessing steps.

```python
from wsdp.processors import BaseProcessor

# Inspect the default processing pipeline
processor = BaseProcessor()
print(f"Processor: {processor}")

# You can also inspect the algorithm registry directly:
from wsdp.algorithms import list_algorithms

print("\nRegistered algorithms:")
for category, methods in list_algorithms().items():
    print(f"  {category}: {', '.join(methods)}")
```
